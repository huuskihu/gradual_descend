## 1 · 变量与类型

```python
# 常见类型，用 type() 查看
a = 784              # int   整数
b = 0.1304           # float 小数
c = "data/mnist.pkl.gz"   # str   字符串
d = True             # bool  真假（首字母必须大写！）
e = [784, 30, 10]    # list  列表，可以改
f = (50000, 784)     # tuple 元组，不能改
g = {"train": 50000, "test": 10000}   # dict 字典：键 -> 值

for name, v in [("a", a), ("b", b), ("c", c), ("d", d), ("e", e), ("f", f), ("g", g)]:
    print(f"{name} = {str(v):25s} 类型是 {type(v).__name__}")

print()
print("取字典里的值:", g["train"])
print("取列表第 1 个:", e[0], "  注意：从 0 开始数！")
```

## 2 · 切片：取一部分

`a[起点:终点]`，**终点不包含**。这是新手最容易记错的地方。

```python
a = [10, 20, 30, 40, 50]
print("a        =", a)
print("a[0]     =", a[0], "   第 1 个（从 0 数起）")
print("a[1:3]   =", a[1:3], "   第 1 到第 3 个【之前】，不含第 3 个")
print("a[:2]    =", a[:2], "   从头到第 2 个之前")
print("a[2:]    =", a[2:], "   从第 2 个到末尾")
print("a[-1]    =", a[-1], "   最后一个（负数 = 从末尾数）")
```

## 3 · f-string：把变量塞进字符串

字符串前加 `f`，变量用 `{}` 包起来。

```python
acc = 95.66123
epoch = 3
print(f"第 {epoch} 轮，准确率 {acc}%")        # 直接塞
print(f"准确率 {acc:.2f}%")                   # .2f = 保留 2 位小数
print(f"准确率 {acc:.0f}%")                   # .0f = 不带小数
print(f"{'及格' if acc > 90 else '不及格'}")   # 还能在 {} 里做判断
```

## 4 · 函数（function）：把一段代码打包

`def 名字(参数):` + 缩进 + `return 结果`

```python
import numpy as np

def sigmoid(z):
    """把任意数字压到 0~1（这是神经网络里最常用的函数之一）"""
    return 1.0 / (1.0 + np.exp(-z))

print("sigmoid(0)   =", sigmoid(0))      # 0.5
print("sigmoid(10)  =", round(float(sigmoid(10)), 4))
print("sigmoid(-10) =", round(float(sigmoid(-10)), 4))
print()
print("一次算三个:", np.round(sigmoid(np.array([-1, 0, 1])), 3))   # NumPy 函数会自动对每个元素分别算
```

## 5 · 类（class）与 self

类 = **图纸**，对象 = **照图纸造的房子**。`self` 就是「我自己」。

```python
class Neuron:
    """一个最简单的神经元：记住自己的权重，能对输入做加权求和"""

    def __init__(self, n_inputs):          # 造房子时自动执行
        self.n = n_inputs                  # 挂在自己身上，以后 self.n 能用
        self.w = np.zeros(n_inputs) + 0.5  # 权重，先都设成 0.5
        self.b = 0.1                       # 偏置

    def forward(self, x):                  # 类里的函数叫"方法"，第一个参数必须是 self
        return x @ self.w + self.b        # 读自己的权重

n = Neuron(4)                              # 造一个能接收 4 个输入的神经元（自动调用 __init__）
print("它有几个输入:", n.n)
print("它的权重    :", n.w)
print("输入 [1,2,3,4] 的输出:", n.forward(np.array([1.0, 2.0, 3.0, 4.0])))
print("手算验证    :", 1*0.5 + 2*0.5 + 3*0.5 + 4*0.5 + 0.1)
```

## 6 · 方法 vs 函数

**东西在前、用点号的是方法；名字在前、括号包东西的是函数。**

```python
x = np.array([[1, 2, 3], [4, 5, 6]])

print("函数 len(x)      =", len(x))
print("方法 x.sum()     =", x.sum())
print("方法 x.shape     =", x.shape, "  <- shape 是属性，不是函数，所以不加括号")
print()
print("dir(x) 能列出 x 所有可用的方法（前 15 个）:")
print([m for m in dir(x) if not m.startswith("_")][:15])
```

## 8 · NumPy：shape、axis、@、.T

这是本项目里出现频率最高的四个概念。

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])           # 2 行 3 列

print("A =
", A)
print("A.shape      =", A.shape, " (2 行, 3 列)")
print()
print("A.sum()      =", A.sum(), "         全部加起来")
print("A.sum(axis=0)=", A.sum(axis=0), "    跨行压扁（上下相加）-> 3 个数")
print("A.sum(axis=1)=", A.sum(axis=1), "    跨列压扁（左右相加）-> 2 个数")
print()
print("A.T =
", A.T, "   转置：行变列、列变行")
```

```python
# 矩阵乘法 @ 与逐元素乘 * 的区别
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[10, 20],
              [30, 40]])

print("A @ B （矩阵乘法，行乘列再相加）=
", A @ B)
print()
print("A * B （对应位置各自相乘）=
", A * B)
print()
print("形状规则: (a,b) @ (b,c) -> (a,c)，中间那个数必须相同")
print("A:", A.shape, " B:", B.shape, " A@B:", (A @ B).shape)
```

```python
# 广播：小数组自动"复制"成一样大再运算
Z = np.array([[1, 2, 3],
              [4, 5, 6]])     # 形状 (2,3)
b = np.array([10, 20, 30])    # 形状 (3,)

print("Z + b =
", Z + b)
print()
print("Z 是 (2,3)，b 是 (3,)，NumPy 自动把 b 复制成 2 份再相加 —— 这叫广播")
print("所以代码里写  X @ W + b  时，b 只有一行也能加，不用手写循环")
```

## 9 · 随机数与种子（seed）

**同样的种子 = 同样的结果**，这样别人才能够复现你的实验。

```python
rng1 = np.random.default_rng(42)     # 种子 42
rng2 = np.random.default_rng(42)     # 还是 42
rng3 = np.random.default_rng(999)    # 换个种子

print("rng1 的前 5 个随机数:", np.round(rng1.normal(size=5), 4))
print("rng2 的前 5 个随机数:", np.round(rng2.normal(size=5), 4), " <- 和 rng1 一模一样！")
print("rng3 的前 5 个随机数:", np.round(rng3.normal(size=5), 4), " <- 换了种子就不同")

# permutation：把顺序打乱（训练时每轮打乱数据用）
rng = np.random.default_rng(0)
print()
print("0~9 打乱后:", rng.permutation(10))
```

## 10 · 实战：用真实数据算一个神经元

```python
import os, sys
from pathlib import Path

# 找到项目根目录（不管你在哪启动 notebook）
cur = Path.cwd()
for p in [cur, *cur.parents]:
    if (p / "data" / "mnist.pkl.gz").exists():
        ROOT = p
        break
else:
    ROOT = cur
os.chdir(ROOT)
sys.path.insert(0, str(ROOT / "src"))

from mnist_data import load_arrays

arrays = load_arrays()
(Xtr, ytr), _, _ = arrays["train"], arrays["val"], arrays["test"]
print("加载成功！训练集:", Xtr.shape, " 第 1 张图的正确答案是:", ytr[0])

# 拿第 1 张图，用一个（随机初始化的）神经元算一次
x = Xtr[0]                                   # 784 个数
rng = np.random.default_rng(0)
w = rng.normal(0, 0.1, size=784)             # 权重：784 个随机数
b = 0.0

z = x @ w + b                                # 加权求和
a = sigmoid(z)                               # 激活

print()
print("x 的形状:", x.shape, " w 的形状:", w.shape)
print("加权求和的结果 z =", round(float(z), 4))
print("经过 sigmoid 后 a =", round(float(a), 4))
print()
print("这就是神经网络里【一个神经元】的完整计算，只有两行。")
```

## 11 · 综合：手写一个 3 行的迷你网络

```python
# 输入 784 -> 隐藏 30 -> 输出 10，随机初始化（还没训练，所以是瞎猜）
rng = np.random.default_rng(42)
W1 = rng.normal(0, np.sqrt(1 / 784), size=(784, 30))
b1 = np.zeros(30)
W2 = rng.normal(0, np.sqrt(1 / 30), size=(30, 10))
b2 = np.zeros(10)

x = Xtr[0:1]                      # 取 1 张图，形状 (1, 784) —— 注意 [0:1] 不是 [0]

Z1 = x @ W1 + b1                  # 第 1 层：加权求和
A1 = np.maximum(0, Z1)            # 第 1 层：ReLU 激活（负数变 0）
Z2 = A1 @ W2 + b2                 # 第 2 层：加权求和

print("形状变化:", x.shape, "->", Z1.shape, "->", Z2.shape)
print()
print("10 个类别的得分:", np.round(Z2[0], 2))
print("它猜的数字是  :", Z2.argmax())
print("正确答案是    :", ytr[0])
print()
print("猜错很正常 —— 权重还是随机的，它什么都没学过。")
print("所谓【训练】，就是不断调整 W1/b1/W2/b2 这四个数组，让它猜对。")
```
